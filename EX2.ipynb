```python
from msilib import add_data

import pandas as pd
data=pd.read_csv('https://raw.githubusercontent.com/Rajvardhan180/House_Price_Prediction/main/data/Housing.csv')
data.head()
```

```python
#Excersize 1
P_data=data[(data["prefarea"]=="yes")&(data["guestroom"]=="yes")&(data["area"]>3000)&(data["airconditioning"]=="yes")]

r=P_data.groupby("furnishingstatus").agg(ave_price=("price","mean"),ave_bedrooms= ("bedrooms","mean"),p_count=("price","count")).reset_index()
print(r)
```

```python
#Excersize 2
data["e_load"]=(data["area"]*0.5+data["bedrooms"]*100+data["bathrooms"]*150)

a_data=data[(data["airconditioning"]=="yes")& (data["hotwaterheating"]=="yes")]

rr=a_data.groupby("furnishingstatus").agg(ave_load=("e_load","mean"),ave_price=("price","mean")).reset_index()
print(rr)
```
