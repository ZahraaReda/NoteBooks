```python
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/mpolinowski/hotel-booking-dataset/refs/heads/master/datasets/hotel_bookings.csv')
df.head()
df.info()
df.describe()

# visualising data to see how it is distributed
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
df.dropna()
# dropping nulls , there is no dups
```

```python
# since i'm using lead_time and previous_cancellations on X axis i dont need to check for outliers so i will skip this part
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd

cols =["is_canceled","lead_time","previous_cancellations"]
df_colss = df[cols]
sns.histplot(data=df[cols],x="lead_time" ,kde=True,bins=50 )
# Histogram for lead time
```

```python
correlation_matrix = df_colss.corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.show()

#heat map
```

```python
x = df[['lead_time', 'previous_cancellations']]
y = df['is_canceled']
x.head()
y.value_counts()
```

```python
y.value_counts(normalize=True)
sns.countplot(x=y)
plt.title('Cancellation Distribution')
plt.show()
# checking to imbalance in is_canceled
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test ,y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42 , stratify=y)
# splitting data to train and test before preprocessing , used stratify to preserve the original proportions
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_tns = sc.fit_transform(x_train)
x_tss= sc.transform(x_test)
# preprocessing data
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_tns, y_train)
# using logistic regression
```

```python
y_p = lr.predict(x_tss)
#predictions
```

```python
from sklearn.metrics import confusion_matrix , ConfusionMatrixDisplay
cm = confusion_matrix(y_test, y_p)
display = ConfusionMatrixDisplay(confusion_matrix=cm)
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Not Canceled', 'Canceled'],
    yticklabels=['Not Canceled', 'Canceled']
)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
#the confusion matrix shows what bookings and correctly classified and what not
```

```python
from sklearn.metrics import precision_score, recall_score
from sklearn.metrics import classification_report
precision = precision_score(y_test, y_p)
recall = recall_score(y_test, y_p)
print(classification_report(y_test, y_p))
print(precision)
print(recall)
```

```python
#since the model is biased towards the 0 class i will apply SMOTE
from imblearn.over_sampling import SMOTE
sm = SMOTE()
xt_s , yt_s = sm.fit_resample(x_tns, y_train)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(yt_s.value_counts())

#after using SMOTE the data is balanced
```

```python
model_smote = LogisticRegression(random_state=42)
model_smote.fit(xt_s, yt_s)

# then prediction
y_ps = model_smote.predict(x_tss)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_ps)

plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Not Canceled', 'Canceled'],
    yticklabels=['Not Canceled', 'Canceled']
)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - SMOTE Logistic Regression')

plt.show()
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_ps))
```

```python
precision_after = precision_score(y_test, y_ps)
recall_after = recall_score(y_test, y_ps)

print("\nAfter SMOTE")
print("Precision:", precision_after)
print("Recall:", recall_after)

# i noticed after SMOTE the model predicts class 1 better but at the same time the accuracy is less
```
